##### 3) Define functions

```python
# Define function combining geopackages
def combine_national_observed_decomposed_metrics(country_list):
    '''
    Function for combining national metrics into one geodataframe.
    Function takes as input:
        - the list of countries to combine (ISO3 codes)
    '''
    # List to store individual GeoDataFrames
    gdfs = []
    # Loop over all countries in the country list
    for country in country_list:
        try:
            metric_path = os.path.join("..", "data", "results", "social_flood", "countries", f"{country}",
                                       "inequality_metrics", f"{country}_ADM0_decomposed_metrics_gfd-flood_S-rwi.gpkg")
            # Read the GeoPackage
            gdf = gpd.read_file(metric_path)
            # Append to list
            gdfs.append(gdf)
        except Exception as e:
            print(f"Error processing {country}: {e}")
    
    # Combine all GeoDataFrames
    if gdfs:
        combined_gdf = pd.concat(gdfs, ignore_index=True)
        return combined_gdf
    else:
        print("No valid data found")

def combine_national_observed_metrics(country_list, flood_type, flood_year, pop_year):
    '''
    Function for combining national metrics into one geodataframe.
    Function takes as input:
        - the list of countries to combine (ISO3 codes)
        - the flood type (inland, coastal, or combined)
        - the flood year (all or 2000-2018)
        - the population year to use (2000-2020)
    '''
    # List to store individual GeoDataFrames
    gdfs = []
    # Loop over all countries in the country list
    for country in country_list:
        try:
            metric_path = os.path.join("..", "data", "results", "social_flood", "countries", f"{country}",
                                       "inequality_metrics", f"{country}_ADM0_metrics_gfd_{flood_year}_{flood_type}-flood_S-rwi_P-{pop_year}.gpkg")
            # Read the GeoPackage
            gdf = gpd.read_file(metric_path)
            # Append to list
            gdfs.append(gdf)
        except Exception as e:
            print(f"Error processing {country}: {e}")
    
    # Combine all GeoDataFrames
    if gdfs:
        combined_gdf = pd.concat(gdfs, ignore_index=True)
        return combined_gdf
    else:
        print("No valid data found")

def name_to_iso3(name):
    try:
        # lookup is case-insensitive and handles many common variants
        return pycountry.countries.lookup(name).alpha_3
    except LookupError:
        return None      # or fallback string
```

```python
# User Inputs
ISO3 = 'IND'
event_id = '4179' # Jhelum River Flooding
# event_id = '3372' # Brahmaputra River Flooding 
```

```python
# File paths
pop_path = os.path.join('..', 'data', 'inputs', 'analysis', 'countries', f'{ISO3}', f'{ISO3}_ghs-pop.tif')
rwi_path = os.path.join('..', 'data', 'inputs', 'analysis', 'countries', f'{ISO3}', f'{ISO3}_rwi.tif')
mask_path = os.path.join('..', 'data', 'inputs', 'analysis', 'countries', f'{ISO3}', f'{ISO3}_surface_water.tif')
flood_path = os.path.join('..', 'data', 'inputs', 'analysis', 'events', f'DFO_{event_id}',f'{ISO3}_{event_id}.tif')
```

```python
# Load raster data
with rasterio.open(rwi_path) as rwi_src, rasterio.open(pop_path) as pop_src, rasterio.open(flood_path) as flood_src, \
    rasterio.open(mask_path) as mask_src:
    rwi = rwi_src.read(1)
    pop = pop_src.read(1)
    mask = mask_src.read(1)
    flood = flood_src.read(1)
    affine = pop_src.transform
rwi[rwi==-999] = np.nan # convert -999 in RWI dataset to NaN
water_mask = np.where(mask>50, np.nan, 1) # WARNING WE ARE HARD CODING PERM_WATER > 50% mask here
```

```python
# Prepare data
# Mask areas 
mask_areas = (
    ~np.isnan(pop) &
    ~np.isnan(rwi) &
    ~np.isnan(flood) &
    ~np.isnan(water_mask)
    )
# Flatten data
pop_flat = pop[mask_areas]
rwi_flat = rwi[mask_areas]
flood_flat = flood[mask_areas]
# Mask out zero-populatoin cells
valid = pop_flat > 0
pop_flat = pop_flat[valid]
rwi_flat = rwi_flat[valid]
flood_flat = flood_flat[valid]
```

```python
# Create analysis dataframe
df = pd.DataFrame({
    'pop': pop_flat,
    'rwi': rwi_flat,
    'flood': flood_flat,
})
```

```python
# Function for extracting concentration curve data
def compute_concentration_curve(data):
    # Sort by RWI ascending
    data = data.sort_values(by="rwi", ascending=True).copy()
    data['cum_pop'] = data['pop'].cumsum()
    data['flood_pop'] = data['flood'] * data['pop']  # flood burden
    data['cum_flood_pop'] = data['flood_pop'].cumsum()
    total_pop = data['pop'].sum()
    total_flood_pop = data['flood_pop'].sum()
    data['frac_pop'] = data['cum_pop'] / total_pop
    data['frac_flood'] = data['cum_flood_pop'] / total_flood_pop
    # Insert 0 at the beginning for a proper starting point at (0,0)
    x_vals = np.insert(data['frac_pop'].values, 0, 0)
    y_vals = np.insert(data['frac_flood'].values, 0, 0)
    return x_vals, y_vals

x, y = compute_concentration_curve(df)
```

```python
# Plot the concentration curve
plt.figure(figsize=(6, 5))
plt.plot(x, y, label=f'DFO {event_id}', color='dodgerblue')
# plt.title('Observed Concentration Curve - Bhutan')
plt.plot([0,1], [0,1], "--", color='gray', label='Equality Line')
plt.xlabel('Cumulative Population (Wealth Rank)')
plt.ylabel('Cumulative Flood Risk Exposure')
plt.legend()
```

```python
def summarise_national_df(df_country: pd.DataFrame) -> pd.DataFrame:
    """
    Input: 1 row per country (or can have duplicates; we collapse by ISO3).
    Output: 1 row per country with exposure totals + quintiles + national CI.
    """
    df = df_country.copy()

    # Ensure numeric columns are numeric
    num_cols = [
        "CI", "Population", "Perc_Exposure",
        "Total Flood Risk", "Q1 Flood Risk", "Q2 Flood Risk", "Q3 Flood Risk", "Q4 Flood Risk", "Q5 Flood Risk",
    ]
    for c in num_cols:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")

    agg_map = {}
    for c in ["Total Flood Risk", "Q1 Flood Risk", "Q2 Flood Risk", "Q3 Flood Risk", "Q4 Flood Risk", "Q5 Flood Risk"]:
        if c in df.columns:
            agg_map[c] = "sum"   # totals across any duplicates
    if "Population" in df.columns:
        agg_map["Population"] = "first"
    if "Perc_Exposure" in df.columns:
        agg_map["Perc_Exposure"] = "first"
    if "CI" in df.columns:
        agg_map["CI"] = "first"
    if "ShapeName" in df.columns:
        agg_map["ShapeName"] = "first"

    out = (
        df.groupby("ISO3", as_index=False)
          .agg(agg_map)
          .rename(columns={
              "ShapeName": "Country",
              "CI": "Total National CI",
              "Total Flood Risk": "Total (Cumulative) Flood Exposure",
          })
    )

    # Relative exposure
    if "Perc_Exposure" in out.columns:
        out["Relative Flood Exposure"] = out["Perc_Exposure"]

    # Quintile exposure outputs
    ren = {
        "Q1 Flood Risk": "Total Quintile Exposure (Q1)",
        "Q2 Flood Risk": "Total Quintile Exposure (Q2)",
        "Q3 Flood Risk": "Total Quintile Exposure (Q3)",
        "Q4 Flood Risk": "Total Quintile Exposure (Q4)",
        "Q5 Flood Risk": "Total Quintile Exposure (Q5)",
    }
    out = out.rename(columns={k: v for k, v in ren.items() if k in out.columns})

    # Columns to keep
    keep = [
        "ISO3", "Country",
        "Total (Cumulative) Flood Exposure",
        "Relative Flood Exposure",
        "Total National CI",
        "Total Quintile Exposure (Q1)",
        "Total Quintile Exposure (Q2)",
        "Total Quintile Exposure (Q3)",
        "Total Quintile Exposure (Q4)",
        "Total Quintile Exposure (Q5)",
    ]
    keep = [c for c in keep if c in out.columns]
    return out[keep]

def summarise_event_df(df_events: pd.DataFrame,
                       iso_col="ISO3",
                       event_id_col="DFO_ID",
                       event_ci_col="CI") -> pd.DataFrame:
    """
    Output columns:
      - Total Number of Events
      - Average Event CI
      - Standard Deviation of Event CI
    """
    df = df_events.copy()
    df[event_ci_col] = pd.to_numeric(df[event_ci_col], errors="coerce")

    if event_id_col in df.columns:
        # count unique events
        n_events = df.groupby(iso_col)[event_id_col].nunique()
    else:
        # fallback: count rows
        n_events = df.groupby(iso_col).size()

    stats = df.groupby(iso_col)[event_ci_col].agg(["mean", "std"])

    out = pd.concat([n_events, stats], axis=1).reset_index()
    out = out.rename(columns={
        iso_col: "ISO3",
        0: "Total Number of Events",
        "mean": "Average Event CI",
        "std": "Standard Deviation of Event CI",
    })

    if "Total Number of Events" not in out.columns:
        out = out.rename(columns={n_events.name: "Total Number of Events"})

    return out
```
